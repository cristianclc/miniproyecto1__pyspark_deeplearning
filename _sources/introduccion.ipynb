# **1. Contexto del Proyecto**

La publicidad digital móvil mueve miles de millones de dólares al año y su eficiencia depende, en gran medida, de una pregunta aparentemente simple: **¿este usuario va a hacer clic en este anuncio?**

Responder esa pregunta en tiempo real, sobre decenas de millones de impresiones diarias, es uno de los problemas más relevantes y exigentes del aprendizaje automático aplicado. No solo importa acertar — importa acertar rápido, a escala, y con datos que en su mayoría son categóricos, anonimizados y altamente desbalanceados hacia la clase negativa (la gran mayoría de los anuncios no reciben clic).

Este proyecto aborda ese desafío usando el dataset **Avazu CTR Prediction**, un benchmark clásico de Kaggle con aproximadamente **40 millones de registros** de impresiones publicitarias móviles reales. El problema se formula como una **clasificación binaria supervisada**: predecir si la variable `click` toma el valor 1 (hubo clic) o 0 (no lo hubo).

El enfoque técnico central es una **Red Neuronal Multicapa (MLP — Multilayer Perceptron)**, implementada y comparada en dos entornos computacionales con filosofías distintas:

- **scikit-learn**: entorno flexible y orientado a experimentación, trabajado sobre una muestra representativa de 1 millón de registros.
- **PySpark**: entorno distribuido orientado a escala, trabajado sobre 10 millones de registros con balanceo de clases mediante *undersampling*.

La comparación entre ambos entornos no es solo técnica — es también práctica: ¿cuándo vale la pena escalar a Spark? ¿Qué se gana y qué se pierde en cada decisión de preprocesamiento? ¿Cómo cambia el comportamiento del modelo según la estrategia de manejo del desbalance?

A estas preguntas se suma una dimensión adicional: la **interpretabilidad**. Mediante **LIME** (*Local Interpretable Model-agnostic Explanations*), se analiza qué variables impulsan o frenan la predicción en instancias individuales, abriendo una ventana hacia la lógica interna del modelo.

## **Base de Datos: Click-Through Rate Prediction**

### **Origen**

El Dataset utilizado en este proyecto corresponde a **Click-Through Rate Prediction**, usado inicialmente en competencias de Kaggle con el fin de evaluar algoritmos para la predicción de clicks en publicidad digital.

### **Características**

* **Volumen:** 40,428,967 registros de usuarios.

* **Variables:** 24 variables.

* **Período:** 11 días de información. 

* **Completitud:** No contamos con valores faltantes en el Dataset.

## **Variables analizadas**

* `id`: Identificador único de cada impresión de anuncio. Cada fila del dataset representa una ocasión en la que un anuncio fue mostrado a un usuario.

* `hour`: Marca temporal del anuncio (formato: YYMMDDHH)

* `banner_pos`: Posición del banner en la página web.

* `site_id`: Identificador único del sitio web donde se mostró el anuncio.

* `site_domain`: Dominio del sitio web donde se muestra el anuncio.

* `site_category`: Categoría temática del sitio web.

* `app_id`: Identificador único de la aplicación móvil donde se mostró el anuncio.

* `app_domain`: Dominio asociado a la aplicación.

* `app_category`: Categoría de la aplicación.

* `device_id`: Identificador único del dispositivo.

* `device_ip`: Dirección IP del dispositivo que vio el anuncio.

* `device_model`: Modelo del dispositivo.

* `device_type`: Tipo de dispositivo.

* `device_conn_type`: Tipo de conexión de red del dispositivo.

* `C1`: variable categórica anonimizada

* `C14–C21`: variables categóricas anónimas

### **Variable Objetivo**

* `click`:  indicador de si hubo clic
